```python
# importing library
import pandas as pd
pd.set_option("display.max_columns", None)
```

```python
#reading listings data
listings = pd.read_csv(r"/kaggle/input/datasets/mysarahmadbhat/airbnb-listings-reviews/Airbnb Data/Listings.csv",encoding="ISO-8859-1",low_memory=False)
listings.info()
```

## Cleaning the Data

```python
# changing data type of "host_since" column to datetime
listings["host_since"] = pd.to_datetime(listings["host_since"])
# viewing the summary statistics of the numerical columns 
listings.describe()
```

```python
# dropping rows with price and accommodates columns having zero values
listings = listings[(listings["price"] != 0) & (listings["accommodates"] != 0)]
listings.shape
```

There were 113 entries with zero values.

```python
# checking for duplicates entries
listings.duplicated().sum()
```

So,there are not any duplicated rows of listings

```python
# city analysis
listings["city"].value_counts()
```

Paris have the most listing entries we will be analyzing paris listings data

## Paris Listings Analysis

```python
# Paris listings
paris_listings = listings[listings["city"] == "Paris"][["listing_id", "host_since","neighbourhood","city","room_type","price", "review_scores_rating","review_scores_cleanliness","review_scores_location"]]
# checking for missing values in Paris listings
paris_listings.isna().sum()
```

```python
# lets drop only host_since null values
paris_listings.dropna(subset=["host_since"], inplace=True)
paris_listings.shape
```

```python
paris_listings.groupby("neighbourhood")["price"].mean().sort_values(ascending=False)
```

We have Elysee most expensive area in Paris with average price of €211 while Menilmontant has the cheapest average rate of €75.

```python
# Yearly insights on listings
paris_listings["year"] = paris_listings["host_since"].dt.year
paris_listings.groupby("year").agg({"listing_id": "count", "price": "mean"}).sort_values("price", ascending=False)
```

**Airbnb Market throughout the years** 
- Paris has lowest listings only 4 which indicates the start of bnb in paris in 2008 which grdaually increases upto 2015 maximum listings of 12147 which shows new hosts joining the bnb rise of bnb market.
- From 2015 paris listings starts decreasing showing a very huge drop due to safety concerns terrorist attacks on Paris.
- Then from 2018 bnb market starts rising again until 2020 and then it hits a massive drop again in 2021 cause of cornona virus and start of pandemic.

__Insights on average Airbnb prices over the years__
- __2008-2009:__ Prices at the start were very low avg of €78 per night but in 2009 prices got almost double €160 per night the highest increase in prices maximum price per night over the years peak of Airbnb market.
- __2010-2014:__ From 2010 prices starts decreasing with avg price falling from €125 to €100 approx.It shows increasing competition as more hosts starts entered the market during those years.
- __2015-2020:__ During those years prices gradually fall and rise until start of corona virus.
- __2021 onwards:__ Start of pandemic ban on tourism with avg of €93 per night biggest drop in prices.

```python
paris_listings.groupby("room_type").agg({"listing_id": "count", "price": "mean", "review_scores_cleanliness": "mean"}).sort_values("price", ascending=False)
```

__Room_type analysis__ 
- We see that almost 80% of the listings are of __Entire place__. 
- As expected hotel rooms are most expensive approx double the price of entire place cause of more services cleanliness rate higher than the others.
- Shared room have least cleanliness rate also the least avg price per night.

```python
paris_listings.groupby("neighbourhood")[["price","review_scores_cleanliness","review_scores_location","review_scores_rating"]].mean().sort_values("review_scores_rating", ascending=False)
```

__Neighbourhood price relation with cleanliness and location__
- __Best neighbourhood__ Hotel-de-Ville has the highest cleanliness and location ratings despite being mid priced.
- Cheap doesn't mean badly Menilmontant has the mid level cleanliness rate while Elysee being the most expensive has low level cleanliness rate and ratings.
- Prices are most likely driven by the location not by the cleanliness.
- Popincourt is the best neighbourhood to live as it has highest ratings.

## Paris Reviews

```python
#reading reviews data 
reviews = pd.read_csv(r"/kaggle/input/datasets/mysarahmadbhat/airbnb-listings-reviews/Airbnb Data/Reviews.csv", encoding="latin-1", low_memory=False)
paris_reviews = pd.merge(paris_listings,reviews, on="listing_id")
#converting date column to datetime format and extracting month from it
paris_reviews["date"] = pd.to_datetime(paris_reviews["date"])
paris_reviews["month"] = paris_reviews["date"].dt.month
paris_reviews.groupby("month")["review_id"].count().sort_values(ascending=False)
```

__Insights on monthly reviews__
- __Jun-Oct:__ Greatest number of reviews during this period showing peak tourism season in Paris leads to more bookings.
- __Feb-April:__ Winter having least number of reviews less tourists interaction during this season.
- Monthly stays could reflect a lag cause reviews are left after their stay so october reviews may be reflecting september stays.
